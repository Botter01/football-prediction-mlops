```python
import pandas as pd
import requests
from io import StringIO

# Premier League adatok
seasons = ["2526", "2425", "2324", "2223"]
base_url = "https://www.football-data.co.uk/mmz4281/{season}/E0.csv"
```

```python
dfs = []
for season in seasons:
    url = base_url.format(season=season)
    response = requests.get(url)
    if response.status_code == 200:
        df = pd.read_csv(StringIO(response.text))
        df["season"] = season
        dfs.append(df)
        print(f"{season} letöltve, {len(df)} sor")
    else:
        print(f"{season} nem elérhető")

raw_data = pd.concat(dfs, ignore_index=True)
raw_data.head()
```

```python
oszlopok = list(raw_data.columns)
oszlopok
```

```python
oszlopok[0] = "Div"
```

```python
raw_data.to_csv("premier_league_raw.csv", index=False)
print(f"Mentve, {len(raw_data)} sor összesen, {len(raw_data.columns)} oszlop")

# Feltöltés Azure Blob Storage-ba
from azureml.core import Workspace

ws = Workspace.from_config()
datastore = ws.get_default_datastore()

datastore.upload_files(
    files=["premier_league_raw.csv"],
    target_path="football-data/raw/",
    overwrite=True
)
print("Feltöltve a Blob Storage-ba")
```

```python
from azure.ai.ml import MLClient
from azure.ai.ml.entities import Data
from azure.ai.ml.constants import AssetTypes
from azure.identity import DefaultAzureCredential

ml_client = MLClient.from_config(DefaultAzureCredential())

data_asset = Data(
    path="azureml://datastores/workspaceblobstore/paths/football-data/raw/premier_league_raw.csv",
    type=AssetTypes.URI_FILE,
    name="premier-league-raw_new",
    version="1",
    description="Premier League meccsadatok 2022-2026, football-data.co.uk-ról"
)
ml_client.data.create_or_update(data_asset)
print("Data Asset regisztrálva")
```
